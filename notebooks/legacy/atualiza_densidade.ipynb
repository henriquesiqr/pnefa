```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
BASE_DIR = Path().resolve().parent
DATA_DIR = BASE_DIR / "data"
```

```python
propriedades_arq = Path(DATA_DIR, "Propriedades_consideradas_FINAL_areas.csv")
```

```python
propriedades_df = pd.read_csv(
    propriedades_arq,
    dtype={
        "codigo_do_municipio": str,
        "codigo_da_propriedade": str,
        "codigo_da_exploracao": str,
        "cpf/cnpj_do_produtor": str,
    }
)
propriedades_df.head()
```

```python
## Preencher dados faltantes de área com a mediana por município
propriedades_df['num_area'] = propriedades_df['num_area'].fillna(
    propriedades_df.groupby('descricao_do_municipio')['num_area'].transform('median')
)
```

```python
propriedades_df['num_area'].isna().sum()
```

```python
# Recalcular densidade de bovinos
propriedades_df['densidade_bovinos'] = propriedades_df['total_de_bovinos'] / propriedades_df['num_area']

# Tratar divisão por zero/infinito
propriedades_df['densidade_bovinos'] = propriedades_df['densidade_bovinos'].replace([np.inf, -np.inf], np.nan)
```

```python
propriedades_df.to_csv("../data/Propriedades_consideradas_FINAL_treated.csv", index=False)
```
