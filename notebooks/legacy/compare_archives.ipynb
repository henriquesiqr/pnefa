```python
import unicodedata

def padronizar_nomes_colunas(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    def normalize(col):
        col = col.lower().replace(' ', '_')
        col = unicodedata.normalize('NFKD', col).encode('ascii', 'ignore').decode('utf-8')
        return col

    df.columns = [normalize(c) for c in df.columns]

    return df
```

```python
df3_pad = padronizar_nomes_colunas(df3)
df4_pad = padronizar_nomes_colunas(df4)
```

```python
def comparar_dfs_ignorando_colunas(df1, df2, colunas_ignorar):
    cols = [
        c for c in df1.columns
        if c in df2.columns and c not in colunas_ignorar
    ]

    print("Colunas só no df1:", sorted(set(df1.columns) - set(df2.columns)))
    print("Colunas só no df2:", sorted(set(df2.columns) - set(df1.columns)))
    print("Comparando colunas comuns:", len(cols))

    a = df1[cols].astype(str).sort_values(by=cols).reset_index(drop=True)
    b = df2[cols].astype(str).sort_values(by=cols).reset_index(drop=True)

    return a.equals(b)
```

```python
comparar_dfs_ignorando_colunas(df3_pad, df4_pad, ["num_area","data_da_atualizacao_cadastral","ano_atualizacao"])
```
