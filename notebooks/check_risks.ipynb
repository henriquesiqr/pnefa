```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(PROJECT_ROOT)# Inspecionar colunas
def checar_coluna(coluna, nome):
    print(f"\n--- {nome} ---")
    
    serie = (
        coluna.astype(str)
        .str.replace(r"\.0$", "", regex=True)
        .str.strip()
        .replace("nan", None)
    )
    
    print("dtype:", coluna.dtype)
    print("tem '.0'? ->", coluna.astype(str).str.contains(r"\.0$").any())
    print("distribuição de tamanho:")
    print(serie.dropna().str.len().value_counts().sort_index())
```

```python
import pandas as pd

df1 = pd.read_csv(PROJECT_ROOT / "data/processed/propriedades_risco_sanitario_legacy.csv")
df2 = pd.read_csv(PROJECT_ROOT / "data/processed/propriedades_risco_sanitario.csv")
```
